Considere o seguinte problema: Temos notas de um grupo de estudantes e essas notas variam de 0 até 100.

 Com base nisso, calcule a média das notas, o número total de elementos da lista, a soma dos valores, o maior valor, o menor valor, a mediana dos valores. (adicionar células explicativas de textos).

notas = [42, 78, 15, 93, 61, 8, 55, 100, 34, 71, 26, 89, 47, 63, 19]

```python
# Calcule o número de elemntos da lista
notas = [42, 78, 15, 93, 61, 8, 55, 100, 34, 71, 26, 89, 47, 63, 19]
print("O número de elementos na lista é:", len(notas))
```

```python
# Calcule a média das notas
print("A média das notas é:", sum(notas)/len(notas))
```

```python
# Calcule a nota máxima
print("A nota máxima é:", max(notas))
```

```python
# Calcule a nota mínima
print("A nota mínima é:", min(notas))
```

```python
# Calcule a mediana
import statistics

mediana = statistics.median(notas)
print("Mediana:", mediana)
```

**Objetivo 2: Organização do código.**

Crie uma função, com o comando def chamada: analisando_dados(lista).

Essa função deve obrigatoriamente receber uma lista como entrada, calcular os mesmo dados calculados no primeiro objetivo, além disso, deve imprimir os resultados de maneira organizada.

```python
import statistics

lista = [42, 78, 15, 93, 61, 8, 55, 100, 34, 71, 26, 89, 47, 63, 19]

def analisando_dados(lista):
  quantidade = len(lista)
  media = sum(lista)/len(lista)
  maximo = max(lista)
  minimo = min(lista)
  mediana = statistics.median(lista)

  print("Quantidade de elementos:", quantidade)
  print("Média:", media)
  print("Máximo:", maximo)
  print("Mínimo:", minimo)
  print("Mediana:", mediana)
analisando_dados(lista)
```

**Objetivo 3: Interpretação dos resultados.**

1. De que forma poderíamos descrever esse conjunto de notas?

É um conjunto de dados que possui 15 notas no total, com muitas variações, onde o menor valor é 8 e o maior é 100.
2. A média consegue representar bem os dados?

O valor da média é 53,4, já o da mediana é 55,0, sendo assim, posso afirmar que a média parece representar bem os dados, pois a proximidade dos valores de média e mediana nos indica que a distribuição dos dados é relativamente simétrica, porque não estão existindo valores extremos que distorçam a interpretação da média.

3. Qual observação relevante podemos trazer a respeito de nossos dados?

A observação seria a de que nossos valores são heterogêneos, ou seja, possuímos valores de notas bem altas e bem baixas.


**Objetivo 4: Crie um gráfico que valide sua afirmação a respeito da média e mediana, um gráfico de distribuição.**

```python
import matplotlib.pyplot as plt

media = sum(notas) / len(notas)
mediana = statistics.median(notas)

plt.hist(notas, bins=8, color="skyblue", edgecolor="black")
plt.axvline(media, color="red", linestyle="--", linewidth=2, label=f"Média: {media:.1f}")
plt.axvline(mediana, color="green", linestyle="--", linewidth=2, label=f"Mediana: {mediana}")

plt.title("Distribuição das notas")
plt.xlabel("Nota")
plt.ylabel("Frequência")
plt.legend()
plt.show()
```

```python
desvio_padrao = statistics.stdev(notas)
print("Desvio padrão:", desvio_padrao)
```

```python
cv = (desvio_padrao / media) * 100
print("Coeficiente de variação:", cv, "%")
```

Calculando o coeficiente de variação (desvio padrão dividido pela média), obtemos aproximadamente 55%. Como valores de CV acima de 30% costumam indicar alta dispersão, isso reforça numericamente que o conjunto de dados é heterogêneo.
